```python
import pandas as pd

df = pd.read_csv("data/cian_all_moscow_full.csv")
df["price_per_m2"] = pd.to_numeric(df["price_per_m2"], errors="coerce")
df["area"] = pd.to_numeric(df["area"], errors="coerce")

# Сколько объявлений на один адрес
addr_counts = df.groupby("address").size()
print(f"Уникальных адресов: {len(addr_counts)}")
print(f"Адресов с 2+ объявлениями: {(addr_counts >= 2).sum()}")
print(f"Адресов с 5+ объявлениями: {(addr_counts >= 5).sum()}")
print()

# Средняя цена за м² по домам (только где 2+ объявления)
house_prices = (df.groupby("address")["price_per_m2"]
                  .agg(["mean", "count"])
                  .rename(columns={"mean": "avg_price_m2", "count": "listings"})
                  .query("listings >= 2")
                  .sort_values("avg_price_m2", ascending=False))

print(f"Домов с 2+ объявлениями: {len(house_prices)}")
print("\nТоп-10 самых дорогих домов:")
print(house_prices.head(10).to_string())
print("\nТоп-10 самых дешёвых домов:")
print(house_prices.tail(10).to_string())
```

```python
import pandas as pd
import numpy as np

df = pd.read_csv("data/cian_all_moscow_full.csv")
df["price"] = pd.to_numeric(df["price"], errors="coerce")
df["price_per_m2"] = pd.to_numeric(df["price_per_m2"], errors="coerce")
df["area"] = pd.to_numeric(df["area"], errors="coerce")

# Курсы (обнови если нужно)
rates = {"rur": 1.0, "usd": 90.0, "eur": 97.0}
df["price_rub"] = df.apply(
    lambda r: r["price"] * rates.get(str(r["currency"]).lower(), 1.0), axis=1
)
df["price_per_m2_rub"] = df.apply(
    lambda r: r["price_per_m2"] * rates.get(str(r["currency"]).lower(), 1.0), axis=1
)

# Фильтруем аномалии — реальный диапазон для Москвы 50к-2млн за м²
df_clean = df[
    (df["price_per_m2_rub"] >= 50_000) &
    (df["price_per_m2_rub"] <= 2_000_000) &
    (df["area"] >= 10) &
    (df["area"] <= 500)
].copy()

print(f"До очистки:  {len(df)}")
print(f"После очистки: {len(df_clean)}")
print(f"Удалено аномалий: {len(df) - len(df_clean)}")

# Средняя цена за м² по домам
house_prices = (df_clean.groupby("address")["price_per_m2_rub"]
                .agg(["mean", "median", "count"])
                .rename(columns={"mean": "avg_m2", "median": "median_m2", "count": "listings"})
                .query("listings >= 2")
                .sort_values("avg_m2", ascending=False)
                .reset_index())

print(f"\nДомов с 2+ объявлениями: {len(house_prices)}")
print("\nТоп-10 дорогих:")
print(house_prices[["address", "avg_m2", "listings"]].head(10).to_string())
print("\nТоп-10 дешёвых:")
print(house_prices[["address", "avg_m2", "listings"]].tail(10).to_string())

# Сохраняем для дальнейшей кластеризации
house_prices.to_csv("house_prices.csv", index=False, encoding="utf-8-sig")
print("\n✅ Сохранено в house_prices.csv")
```
